```python
from zeroconf import Zeroconf, ServiceInfo, ServiceBrowser, ServiceBrowser, ServiceListener
import socket
import hashlib
import time

class DesktopDiscover(ServiceListener):
    def __init__(self):
        self.zc = Zeroconf()
        self.discovered_vaults = {}
        self.service_type = "_securevault._tcp.local."
    
    @staticmethod
    def get_local_ipv4():
        s = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
        try:
            s.connect(('8.8.8.8', 1))  #Google's DNS IP(any connection outside LAN) 
            ip = s.getsockname()[0]
        except:
            ip = 'ERROR'
        finally:
            s.close()
        return ip
    
    def broadcast_presence(self, dsa_pub_key: bytes):
        BROADCASTER_DEVICE = socket.gethostname()


        sha_object = hashlib.sha3_256(dsa_pub_key)
        hash_dsa_pubk = sha_object.hexdigest()
        
        
        LOCAL_IP = self.get_local_ipv4()
        info = ServiceInfo(
            type_ = self.service_type,
            name = f"{BROADCASTER_DEVICE}.{self.service_type}",
            addresses = [socket.inet_aton(LOCAL_IP)],
            port = 65432,
            properties = {"ML-DSA-HASH" : hash_dsa_pubk},
            server = f"{BROADCASTER_DEVICE}.local."
        )
        

        self.zc.register_service(info)
    
    def scan_for_devices(self, scan_time_seconds = 5):
        self.browser = ServiceBrowser(self.zc, self.service_type, self)
        time.sleep(scan_time_seconds)
        self.browser.cancel()
        return self.discovered_vaults
    
    def add_service(self, zc : Zeroconf, type_ : str, name : str):
        info = zc.get_service_info(type_,name)
        if info and info.addresses:
            ip_address = socket.inet_ntoa(info.addresses[0])
            dsa_hash = info.properties.get("ML-DSA-HASH", b"").decode()

            self.discovered_vaults[name] = {
                "ip" : ip_address,
                "port" : info.port,
                "ML-DSA-HASH" : dsa_hash    
            }
    
    def update_service(self, zc, type_, name):
        pass
    
    def remove_service(self, zc, type_, name):
        if name in self.discovered_vaults:
            del self.discovered_vaults[name]
    
    def cleanup(self):
        if self.broadcasting_info:
            self.zc.unregister_service(self.broadcasting_info)
        self.zc.close()
```

```python
import qrcode
import socket


# Your inputs
ipv4_str = "192.168.1.1"
sha256_hex = "e3b0c44298fc1c149afbf4c8996fb92427ae41e4649b934ca495991b7852b855"

# 1. Convert IP to 4-byte binary stream
ip_bytes = socket.inet_aton(ipv4_str)  # Length: 4 bytes

# 2. Convert HEX string to 32-byte binary stream
hash_bytes = bytes.fromhex(sha256_hex)  # Length: 32 bytes

# 3. Concatenate safely
combined_bytes = ip_bytes + hash_bytes

# 1. Create a QRCode object with optimized settings
qr = qrcode.QRCode(
    version=None,  # Automatically detects the smallest possible size
    error_correction=qrcode.constants.ERROR_CORRECT_L,  # Low error correction (best for maximum data capacity)
    box_size=10,  # Size of each square pixel
    border=4,  # Standard border thickness
)

# 2. Add the raw byte data directly
# Note: Do not convert bytes to a string; pass them as-is
qr.add_data(combined_bytes)
qr.make(fit=True)

# 3. Generate and save the image
img = qr.make_image(fill_color="black", back_color="white")
print(img)
```

```python
import platform
import subprocess
import json

def get_friendly_device_name():
    os_type = platform.system()
    
    # --- WINDOWS CONFIGURATION ---
    if os_type == "Windows":
        try:
            # Query the motherboard manufacturer and product name
            cmd = "powershell -Command \"Get-CimInstance Win32_ComputerSystemProduct | Select-Object Vendor, Name | ConvertTo-Json\""
            output = subprocess.check_output(cmd, shell=True, stderr=subprocess.DEVNULL).decode('utf-8')
            data = json.loads(output)
            vendor = data.get('Vendor', '').replace('Inc.', '').strip()
            name = data.get('Name', '').strip()
            return f"{vendor} {name}"
        except Exception:
            return "Windows PC"

    # --- MACOS CONFIGURATION ---
    elif os_type == "Darwin":
        try:
            # Extract official marketing name from the Apple system registry
            cmd = "ioreg -l | grep -o '\"product-description\" = <\".*\">' | head -1"
            output = subprocess.check_output(cmd, shell=True, stderr=subprocess.DEVNULL).decode('utf-8')
            if output:
                # Decode hex string inside the registry angle brackets
                hex_str = output.split('<"')[1].split('">')[0]
                return bytes.fromhex(hex_str).decode('utf-8').strip()
        except Exception:
            pass
        
        try:
            # Fallback to model identifier (e.g., MacBookPro18,3) if marketing name fails
            cmd_fallback = "sysctl -n hw.model"
            return subprocess.check_output(cmd_fallback, shell=True).decode('utf-8').strip()
        except Exception:
            return "Mac Device"

    # --- LINUX CONFIGURATION ---
    elif os_type == "Linux":
        try:
            # Linux reads device names directly from virtual system files
            with open("/sys/class/dmi/id/product_name", "r") as f:
                product = f.read().strip()
            with open("/sys/class/dmi/id/sys_vendor", "r") as f:
                vendor = f.read().strip()
            return f"{vendor} {product}"
        except Exception:
            return "Linux Device"

    return "Unknown Device"


get_friendly_device_name()

```
